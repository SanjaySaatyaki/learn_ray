```python
def expensive_sq(x):
    time.sleep(5)
    return x **2
```

```python
@ray.remote
def remote_expensive_sq(x):
    time.sleep(5)
    return x **2
```

```python
results = []
for item in range(4):
    result = ray.get(remote_expensive_sq.remote(item)) # here things are called mulitple times sequentially
    results.append(result)

print(result)
```
